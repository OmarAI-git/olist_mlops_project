### Create labeled data table

```python
labeled_data = data[data['order_delivered_customer_date'].notna()].copy()
```

```python
labeled_data
```

### Create the late feature

```python
labeled_data['late'] = (labeled_data['order_delivered_customer_date'] > labeled_data['order_estimated_delivery_date']).astype(int)
```

```python
labeled_data.head(10)
```

```python
labeled_data.loc[:,
                  ['order_delivered_customer_date',
                  'order_estimated_delivery_date',
                  'late']
                ].sample(10)
```

```python
labeled_data['late'].value_counts(normalize = True) * 100
```

```python
values = labeled_data['late'].value_counts()
labels = ['on time', 'late']
colors = ['skyblue', 'aqua']
explode = [0.1, 0.2]

plt.pie(x =values, labels = labels,colors = colors, explode = explode, shadow = True, autopct = '%0.2f%%')

plt.show()

# We have class imbalance here
```
