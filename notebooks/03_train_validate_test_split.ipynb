### Split the data

```python
train = data[data['order_purchase_timestamp'] < '2018-04-15 20:09:09.500'].copy()

validation = data[(data['order_purchase_timestamp'] >= '2018-04-15 20:09:09.500') & (data['order_purchase_timestamp'] <  '2018-06-21 08:19:46.500')].copy()

test = data[data['order_purchase_timestamp'] >= '2018-06-21 08:19:46.500'].copy()
```

```python
print(f'Total: {len(data)}')
print(f'Train: {len(train)}')
print(f'Validation: {len(validation)}')
print(f'Test: {len(test)}')

print(f'Sum: {len(train) + len(validation) + len(test)}')
```

```python
print(f'Train:\n')
print(train['order_purchase_timestamp'].min())
print(train['order_purchase_timestamp'].max())


print(f'\nValidation:\n')
print(validation['order_purchase_timestamp'].min())
print(validation['order_purchase_timestamp'].max())


print(f'\nTest:\n')
print(test['order_purchase_timestamp'].min())
print(test['order_purchase_timestamp'].max())

```

```python
print(f'Train:\n')
print(train['late'].value_counts(normalize = True) * 100)

print(f'\nValidation:\n')
print(validation['late'].value_counts(normalize = True) * 100)

print(f'\nTest:\n')
print(test['late'].value_counts(normalize = True) * 100)
```

### Save split files

```python
train.to_parquet('../data/artifacts/03_train.parquet')
validation.to_parquet('../data/artifacts/03_validation.parquet')
test.to_parquet('../data/artifacts/03_test.parquet')
```
